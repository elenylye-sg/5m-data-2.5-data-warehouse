### Question 1


Let's revisit our `austin_bikeshare_demo` dbt project. Modify the `dim_station.sql` model to include the following columns:

- `avg_duration` (Average of trip durations for trips starting from each station in seconds.)

Then, rebuild the models with the following command to see if the changes are correct:

```bash
dbt run
```


**Answer:**

Paste the `dim_station.sql` model here:

```sql
-- Enter Solution here
{{ config(materialized='table') }}

with stations as (
    select
        safe_cast(station_id as string) as station_id,
        name as station_name,
        status,
        address,
        property_type,
        number_of_docks,
        council_district,
        modified_date
    from {{ source('austin_bikeshare', 'bikeshare_stations') }}
    where station_id is not null
),

station_trip_metrics as (
    select
        safe_cast(start_station_id as string) as station_id,
        round(avg(duration_minutes * 60), 2) as avg_duration
    from {{ source('austin_bikeshare', 'bikeshare_trips') }}
    where start_station_id is not null
      and duration_minutes > 0
    group by 1
)

select
    s.station_id,
    s.station_name,
    s.status,
    s.address,
    s.property_type,
    s.number_of_docks,
    s.council_district,
    s.modified_date,
    t.avg_duration
from stations s
left join station_trip_metrics t
    on s.station_id = t.station_id

```
